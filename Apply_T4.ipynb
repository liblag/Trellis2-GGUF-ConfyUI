# Профили T4 и исправление удаления фона
Остановите ComfyUI, сохранив среду. Выполните ячейку и загрузите Trellis2-T4.zip. Затем снова запустите ComfyUI.

```python
from google.colab import files
from pathlib import Path
import zipfile, subprocess, sys
uploaded = files.upload()
archive = next((Path(n) for n in uploaded if n.endswith('.zip')), None)
if archive is None:
    raise ValueError('Upload Trellis2-T4.zip')
dest = Path('/content/Trellis2-T4')
with zipfile.ZipFile(archive) as z:
    for item in z.infolist():
        target = (dest / item.filename).resolve()
        if not target.is_relative_to(dest.resolve()):
            raise ValueError('Unsafe archive path')
    z.extractall(dest)
py = '/content/venv312/bin/python'
constraints = Path('/content/torch_constraints.txt')
if not Path(py).exists() or not constraints.exists():
    raise RuntimeError('Expected existing venv312 and torch_constraints.txt; run this in your original Colab runtime.')
subprocess.run([py, '-m', 'pip', 'install', 'rembg[cpu]', '-c', str(constraints)], check=True)
subprocess.run([py, str(dest/'install_patch.py')], check=True)
print('Ready. Restart the ComfyUI launch cell; import JSON files from /content/Trellis2-T4/example_workflows.')
```
